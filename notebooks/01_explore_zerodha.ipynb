```python
import os
from datetime import datetime
from dotenv import load_dotenv
from kiteconnect import KiteConnect

load_dotenv()
```

```python
api_key = os.getenv("KITE_API_KEY")
access_token = os.getenv("KITE_ACCESS_TOKEN")

missing = [
    name for name, value in [
        ("KITE_API_KEY", api_key),
        ("KITE_ACCESS_TOKEN", access_token),
    ]
    if not value
]
if missing:
    raise RuntimeError(
        f"Missing environment variable(s): {', '.join(missing)}. "
        "Set them before running this cell."
    )
```

```python
kite = KiteConnect(api_key=api_key)
kite.set_access_token(access_token)

# Look up the instrument token (this example searches NSE for INFY).
instrument = next(
    item for item in kite.instruments("NSE")
    if item["tradingsymbol"] == "INFY"
)
kite.set_access_token(os.environ["KITE_ACCESS_TOKEN"])

# Look up the instrument token (this example searches NSE for INFY).
instrument = next(
    item for item in kite.instruments("NSE")
    if item["tradingsymbol"] == "INFY"
)
```

```python
candles = kite.historical_data(
    instrument_token=instrument["instrument_token"],
    from_date=datetime(2025, 1, 1, 9, 15),
    to_date=datetime(2025, 1, 2, 15, 30),
    interval="day",
)

for candle in candles:
    print(candle)
```
